```python
from langgraph.graph import MessagesState, START, END
from langgraph.graph import StateGraph
def make_graph():
    memory = MemorSaver()
    builder = StateGraph(MessagesState)
    builder.add_edge(START, "supervisor")
    builder.add_node("supervisor", supervisor_node)
    builder.add_node(f"{math_node.keywords["name"]}", math_node)
    builder.add_node(f"{poem_node.keywords["name"]}", poem_node)
    builder.add_edge(f"{math_node.keywords["name"]}", "supervisor")
    builder.add_edge(f"{poem_node.keywords["name"]}", "supervisor")
    #builder.add_node("web_researcher", web_research_node)
    graph = builder.compile(interrupt_after=[f"{poem_node.keywords["name"]}"], checkpointer=memory)
    return graph
```
